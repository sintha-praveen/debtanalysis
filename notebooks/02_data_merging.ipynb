```python
import os
import pandas as pd

PROCESSED_DIR = os.path.join("..", "data", "processed")

print("=== Starting Step 2: Data Merging & Filtering ===")

fact_path = os.path.join(PROCESSED_DIR, "cleaned_IDS_ALLCountries_Data_long.csv")
country_path = os.path.join(PROCESSED_DIR, "cleaned_IDS_CountryMetaData.csv")
series_path = os.path.join(PROCESSED_DIR, "cleaned_IDS_SeriesMetaData.csv")

print("Loading cleaned processed datasets...")
df_fact = pd.read_csv(fact_path)
df_country = pd.read_csv(country_path)
df_series = pd.read_csv(series_path)

print(f"Fact Table rows: {df_fact.shape[0]}")
print(f"Country Metadata rows: {df_country.shape[0]}")
print(f"Series Metadata rows: {df_series.shape[0]}")

# 1. Select specific analytical dimensions
df_country_sub = df_country[['Code', 'Region', 'Income Group']].copy()
df_country_sub.rename(columns={'Code': 'Country Code', 'Income Group': 'Income_Group'}, inplace=True)

df_series_sub = df_series[['Code', 'Topic']].copy()
df_series_sub.rename(columns={'Code': 'Series Code'}, inplace=True)

# 2. Merge Fact table with Country and Series metadata using Left Joins
print("\nJoining fact table with dimensional attributes...")
df_merged = df_fact.merge(df_country_sub, on='Country Code', how='left')
df_merged = df_merged.merge(df_series_sub, on='Series Code', how='left')

# 3. Handle unclassified/unmapped values
df_merged['Region'] = df_merged['Region'].fillna('Unknown')
df_merged['Income_Group'] = df_merged['Income_Group'].fillna('Unknown')
df_merged['Topic'] = df_merged['Topic'].fillna('General / IMF')

# 4. Save consolidated analytical file
merged_out = os.path.join(PROCESSED_DIR, "merged_international_debt_analysis.csv")
df_merged.to_csv(merged_out, index=False)

print(f"\nSaved consolidated table: {merged_out}")
print(f"Final Merged Shape: {df_merged.shape}")
print(f"Remaining Missing Values: {df_merged.isnull().sum().sum()}")
print("=== Step 2 Finished Successfully! ===")

df_merged.head()
```
