```python
import pandas as pd
from sqlalchemy import create_engine, text

# Database connection
engine = create_engine(
    "postgresql+psycopg2://postgres:admin@localhost:5432/international_debt_db"
)

# 30 SQL Queries matching exact schema
queries = {
    # 1-5: Overview & Validation
    "Q1": 'SELECT COUNT(*) AS total_countries FROM countries;',
    "Q2": 'SELECT COUNT(*) AS total_indicators FROM indicators;',
    "Q3": 'SELECT COUNT(*) AS total_debt_records FROM debt_data;',
    "Q4": 'SELECT ROUND(SUM("Debt_Value_USD")::numeric, 2) AS total_global_debt FROM debt_data;',
    "Q5": 'SELECT COUNT(DISTINCT "Country Code") AS unique_countries FROM debt_data;',

    # 6-10: Distributions & Statistical Spread
    "Q6": 'SELECT COUNT(DISTINCT "Series Code") AS unique_indicators FROM debt_data;',
    "Q7": 'SELECT MIN("Year") AS min_year, MAX("Year") AS max_year FROM debt_data;',
    "Q8": 'SELECT ROUND(AVG("Debt_Value_USD")::numeric, 2) AS avg_debt FROM debt_data WHERE "Debt_Value_USD" > 0;',
    "Q9": 'SELECT ROUND(MAX("Debt_Value_USD")::numeric, 2) AS max_debt FROM debt_data;',
    "Q10": 'SELECT ROUND(MIN("Debt_Value_USD")::numeric, 2) AS min_debt FROM debt_data WHERE "Debt_Value_USD" > 0;',

    # 11-15: Country & Time Series Trends
    "Q11": 'SELECT "Country", ROUND(SUM("Debt_Value_USD")::numeric, 2) AS total_debt FROM debt_data GROUP BY "Country" ORDER BY total_debt DESC LIMIT 10;',
    "Q12": 'SELECT "Country", ROUND(SUM("Debt_Value_USD")::numeric, 2) AS total_debt FROM debt_data GROUP BY "Country" HAVING SUM("Debt_Value_USD") > 0 ORDER BY total_debt ASC LIMIT 10;',
    "Q13": 'SELECT "Year", ROUND(SUM("Debt_Value_USD")::numeric, 2) AS yearly_debt FROM debt_data GROUP BY "Year" ORDER BY "Year";',
    "Q14": 'SELECT "Year", ROUND(SUM("Debt_Value_USD")::numeric, 2) AS peak_debt FROM debt_data GROUP BY "Year" ORDER BY peak_debt DESC LIMIT 1;',
    "Q15": 'SELECT "Year", COUNT(*) AS record_count FROM debt_data GROUP BY "Year" ORDER BY "Year";',

    # 16-20: Indicators & Metrics
    "Q16": 'SELECT "Indicator", ROUND(SUM("Debt_Value_USD")::numeric, 2) AS total_amount FROM debt_data GROUP BY "Indicator" ORDER BY total_amount DESC LIMIT 10;',
    "Q17": 'SELECT "Indicator", COUNT(*) AS frequency FROM debt_data GROUP BY "Indicator" ORDER BY frequency DESC LIMIT 10;',
    "Q18": 'SELECT "Indicator", ROUND(AVG("Debt_Value_USD")::numeric, 2) AS avg_amount FROM debt_data GROUP BY "Indicator" ORDER BY avg_amount DESC LIMIT 10;',
    "Q19": 'SELECT "Country", COUNT(DISTINCT "Series Code") AS indicator_count FROM debt_data GROUP BY "Country" ORDER BY indicator_count DESC LIMIT 10;',
    "Q20": 'SELECT "Country", "Indicator", "Year", "Debt_Value_USD" FROM debt_data ORDER BY "Debt_Value_USD" DESC LIMIT 5;',

    # 21-25: Regional & Socioeconomic Grouping
    "Q21": 'SELECT COALESCE("Region", \'Unknown\') AS region, ROUND(SUM("Debt_Value_USD")::numeric, 2) AS total_debt FROM debt_data GROUP BY "Region" ORDER BY total_debt DESC;',
    "Q22": 'SELECT COALESCE("Income Group", \'Unknown\') AS income_group, ROUND(SUM("Debt_Value_USD")::numeric, 2) AS total_debt FROM debt_data GROUP BY "Income Group" ORDER BY total_debt DESC;',
    "Q23": 'SELECT "Region", COUNT(DISTINCT "Country Code") AS countries_count FROM debt_data GROUP BY "Region";',
    "Q24": 'SELECT "Income Group", COUNT(*) AS countries_count FROM countries GROUP BY "Income Group";',
    "Q25": 'SELECT "Country", ROUND(SUM("Debt_Value_USD")::numeric, 2) AS debt_latest FROM debt_data WHERE "Year" = (SELECT MAX("Year") FROM debt_data) GROUP BY "Country" ORDER BY debt_latest DESC LIMIT 10;',

    # 26-30: Window Functions & Advanced Aggregations
    "Q26": 'SELECT "Year", ROUND(SUM("Debt_Value_USD")::numeric, 2) AS current_year_debt, ROUND(LAG(SUM("Debt_Value_USD")) OVER (ORDER BY "Year")::numeric, 2) AS prev_year_debt FROM debt_data GROUP BY "Year" ORDER BY "Year";',
    "Q27": 'SELECT "Country", ROUND(SUM("Debt_Value_USD")::numeric, 2) AS total_debt, DENSE_RANK() OVER (ORDER BY SUM("Debt_Value_USD") DESC) AS rank FROM debt_data GROUP BY "Country" ORDER BY rank LIMIT 10;',
    "Q28": 'SELECT "Year", ROUND(AVG(SUM("Debt_Value_USD")) OVER (ORDER BY "Year" ROWS BETWEEN 2 PRECEDING AND CURRENT ROW)::numeric, 2) AS rolling_3yr_avg FROM debt_data GROUP BY "Year" ORDER BY "Year";',
    "Q29": 'SELECT COUNT(*) AS zero_debt_rows FROM debt_data WHERE "Debt_Value_USD" = 0;',
    "Q30": 'SELECT "Year", ROUND(SUM(SUM("Debt_Value_USD")) OVER (ORDER BY "Year")::numeric, 2) AS cumulative_running_total FROM debt_data GROUP BY "Year" ORDER BY "Year";'
}

def execute_query(qn):
    qn = qn.upper()
    if qn in queries:
        print(f"\n--- Running {qn} ---")
        with engine.connect() as conn:
            df = pd.read_sql(text(queries[qn]), conn)
            print(df)
    else:
        print(f"❌ {qn} not found! Use Q1 to Q30.")

# Example execution:
execute_query("Q4")
```
